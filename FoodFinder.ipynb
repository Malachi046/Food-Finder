```python
import pandas as pd
import numpy as np

# Load the dataset
data = pd.read_csv('ForagingClusters.csv') # col -> lat,lng

print("Enter Latitude: ") #[26,49]
lat = input()
print("Enter Longitude: ") #[-67,-124]
lon = input()

lat = float(lat)
lon = float(lon)
```

```python
from math import sqrt
smallest = 99999
cluster = 0

if(lat>10)&(lat<70)&(lon<-50)&(lon>-140):
  for i , value in enumerate(data['lat']):
    a = (lat-data['lat'][i])**2
    b = (lon-data['lng'][i])**2
    c = sqrt(a+b)
    if(c<smallest):
      smallest = c
      cluster = i
else:
  print("Coordinates Out of Range")

print(cluster+1)

```

```python
a = (lat-data['lat'][cluster])
b = (lon-data['lng'][cluster])

if(a < 0):
  print('go south',round(abs(a)*69,ndigits=1),' miles')
else:
  print('go north', round(a*69,ndigits=1),' miles')

if(b < 0):
  print('go east',round(abs(b)*69,ndigits=1),' miles')
else:
  print('go west',round(b*69,ndigits=1),' miles')
```

```python
min = 99999
food = 0

data2 = pd.read_csv('ForagingData.csv') # col -> lat,lng,cluster

for j, location in enumerate(data2['cluster']):
    if(float(location)+1 == cluster-1):
        print(data2['lat'][j],data2['lng'][j],data2['name'][j])
        a = (lat-data2['lat'][j])**2
        b = (lon-data2['lng'][j])**2
        c = sqrt(a+b)
        if(c<min):
          min = c
          food = j


```

```python
print('for the nearest food (',data2['name'][food],')')

A = (lat-data2['lat'][food])
B = (lon-data2['lng'][food])

if(A < 0):
  print('go south',round(abs(A)*69,ndigits=1),' miles')
else:
  print('go north', round(A*69,ndigits=1),' miles')

if(B < 0):
  print('go east',round(abs(B)*69,ndigits=1),' miles')
else:
  print('go west',round(B*69,ndigits=1),' miles')
```
